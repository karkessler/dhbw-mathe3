# 1 · Wärmeleitung als Modellproblem

Wir betrachten die eindimensionale Wärmeleitung

$$u_t = \kappa u_{xx}, \qquad 0<x<1,$$

mit $u(0,t)=u(1,t)=0$. Als Anfangswert verwenden wir

$$u(x,0)=\sin(\pi x)+\tfrac14\sin(3\pi x).$$

Beide Sinusmoden erfüllen die Randbedingungen. Deshalb ist die exakte Lösung bekannt und eignet sich hervorragend zur Kontrolle eines numerischen Verfahrens.

```python
import sys
from pathlib import Path

import matplotlib.pyplot as plt
import numpy as np

project = Path.cwd()
if not (project / 'galerkin_heat.py').exists() and (project.parent / 'galerkin_heat.py').exists():
    project = project.parent
if not (project / 'galerkin_heat.py').exists():
    from urllib.request import urlretrieve
    source = 'https://raw.githubusercontent.com/karkessler/dhbw-mathe3/main/galerkin/galerkin_heat.py'
    urlretrieve(source, project / 'galerkin_heat.py')
sys.path.insert(0, str(project))

from galerkin_heat import exact_solution
```

```python
def plot_exact(time=0.05, diffusivity=1.0):
    x = np.linspace(0.0, 1.0, 401)
    plt.figure(figsize=(8, 4))
    plt.plot(x, exact_solution(x, time, diffusivity), lw=2)
    plt.xlabel('Ort x')
    plt.ylabel('Temperatur u(x,t)')
    plt.title(f'Exakte Lösung bei t={time:.3f}, κ={diffusivity:.2f}')
    plt.grid(alpha=0.3)
    plt.ylim(-0.3, 1.3)
    plt.show()

plot_exact()
```

## Interaktiv experimentieren

Verändere Zeit und Diffusivität. Eine größere Diffusivität lässt hohe Frequenzen besonders schnell abklingen.

```python
try:
    from ipywidgets import FloatSlider, interact
    interact(
        plot_exact,
        time=FloatSlider(value=0.05, min=0.0, max=0.3, step=0.01),
        diffusivity=FloatSlider(value=1.0, min=0.1, max=2.0, step=0.1),
    )
except ImportError:
    print('Für Regler bitte ipywidgets installieren. Die statische Grafik funktioniert bereits.')
```
