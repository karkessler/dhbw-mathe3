# 2 · Galerkin-Verfahren und Trapezregel

Wir zerlegen das Intervall in gleich große Elemente und verwenden stückweise lineare Hutfunktionen $\varphi_i$. Der Ansatz lautet

$$u_h(x,t)=\sum_i c_i(t)\varphi_i(x).$$

Nach Multiplikation mit Testfunktionen und partieller Integration entsteht

$$M\dot c(t)+\kappa Kc(t)=0.$$

$M$ ist die Massenmatrix und $K$ die Steifigkeitsmatrix. Für den Zeitschritt verwenden wir Crank–Nicolson, also die Trapezregel:

$$(M+\tfrac{\Delta t\kappa}{2}K)c^{n+1}=(M-\tfrac{\Delta t\kappa}{2}K)c^n.$$

```python
import sys
from pathlib import Path

import matplotlib.pyplot as plt

project = Path.cwd()
if not (project / 'galerkin_heat.py').exists() and (project.parent / 'galerkin_heat.py').exists():
    project = project.parent
if not (project / 'galerkin_heat.py').exists():
    from urllib.request import urlretrieve
    source = 'https://raw.githubusercontent.com/karkessler/dhbw-mathe3/main/galerkin/galerkin_heat.py'
    urlretrieve(source, project / 'galerkin_heat.py')
sys.path.insert(0, str(project))

from galerkin_heat import exact_solution, l2_error, solve_heat_equation
```

```python
def compare(elements=30, dt=0.004, final_time=0.1):
    solution = solve_heat_equation(elements, dt, final_time)
    reference = exact_solution(solution.x, final_time)

    plt.figure(figsize=(8, 4))
    plt.plot(solution.x, reference, '--', lw=2, label='exakt')
    plt.plot(solution.x, solution.values[-1], 'o-', ms=3, label='Galerkin')
    plt.xlabel('Ort x')
    plt.ylabel('u(x,t)')
    plt.title(f'{elements} Elemente, Δt={dt:g}, Fehler={l2_error(solution):.2e}')
    plt.grid(alpha=0.3)
    plt.legend()
    plt.show()

compare()
```

## Gitter und Zeitschritt verändern

Mehr Elemente verbessern die räumliche Auflösung. Ein kleineres $\Delta t$ reduziert den Zeitdiskretisierungsfehler.

```python
try:
    from ipywidgets import FloatLogSlider, IntSlider, interact
    interact(
        compare,
        elements=IntSlider(value=30, min=6, max=80, step=2),
        dt=FloatLogSlider(value=0.004, base=10, min=-3.0, max=-1.3, step=0.1),
        final_time=FloatLogSlider(value=0.1, base=10, min=-2.0, max=-0.5, step=0.1),
    )
except ImportError:
    print('Für Regler bitte ipywidgets installieren.')
```
